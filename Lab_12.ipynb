# MATRIX CHAIN MULTIPLICATION Dynamic Programming with table filling.
Problem Description:
Find minimum number of scalar multiplications needed to multiply a chain of matrices..

#Theory
Cost of multiplying two matrices:
If A is p×q and B is q×r, cost = p × q × r

Dynamic Programming idea:
Break problem into smaller subproblems.
Store results in a table to avoid recomputation.

m[i][j] = minimum cost to multiply Ai to Aj

## Algorithm
1. Take array p[] of size n
2. Create table m[n][n]
3. Set m[i][i] = 0
4. Fill table diagonally
5. For each chain length L:
    Compute m[i][j]
    Try all possible k between i and j
6. Choose minimum cost
7. Return m[1][n-1]

```python
import sys

def matrix_chain_multiplication(p):
    n = len(p)

    # Create DP table for costs
    m = [[0 for _ in range(n)] for _ in range(n)]
    # Create DP table for split points (optimal parenthesization)
    s = [[0 for _ in range(n)] for _ in range(n)] # To store k values

    # L is chain length
    for L in range(2, n):
        for i in range(1, n - L + 1):
            j = i + L - 1
            m[i][j] = sys.maxsize

            for k in range(i, j):
                cost = (m[i][k] +
                        m[k+1][j] +
                        p[i-1] * p[k] * p[j])

                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k # Store the split point

    return m, s
```

```python
# Input dimensions
p = [10, 30, 5, 60]

m, s = matrix_chain_multiplication(p)

print("Minimum multiplication cost:", m[1][len(p)-1])

print("\nDP Table (costs):")
for row in m[1:]:
    print(row[1:])

print("\nDP Table (split points):")
for row in s[1:]:
    print(row[1:])
```

# Exercise
1. Given p = [5, 10, 3, 12, 5, 50, 6]
2. Compute minimum cost
3. Print optimal parenthesization
4. Compare number of multiplications saved

```python
p = [5,10,3,12,5,50,6]

table = matrix_chain_multiplication(p)

print("Minimum multiplication cost:", table[1][len(p)-1])

print("\nDP Table:")
for row in table[1:]:
    print(row[1:])
```

```python
def print_optimal_parenthesization(s, i, j):
    if i == j:
        return f"A{i}"
    else:
        k = s[i][j]
        left_part = print_optimal_parenthesization(s, i, k)
        right_part = print_optimal_parenthesization(s, k + 1, j)
        return f"({left_part}{right_part})"

def naive_multiplication_cost(p):
    n = len(p)
    if n <= 2:
        return 0
    cost = 0
    current_p_rows = p[0]
    for i in range(1, n - 1):
        cost += current_p_rows * p[i] * p[i+1]
        current_p_rows = p[0] # The resulting matrix always has the first dimension p[0]
    return cost


p = [5, 10, 3, 12, 5, 50, 6]

m, s = matrix_chain_multiplication(p)

optimal_cost = m[1][len(p)-1]
print(f"Minimum multiplication cost: {optimal_cost}")

print("Optimal Parenthesization:", print_optimal_parenthesization(s, 1, len(p) - 1))

# Calculate naive cost for comparison
naive_cost = naive_multiplication_cost(p)
print(f"Naive (left-to-right) multiplication cost: {naive_cost}")

saved_multiplications = naive_cost - optimal_cost
print(f"Multiplications saved: {saved_multiplications}")
```
