# 1. Growth Of Function
Time Complexity/Order of Growth defines the amount of time taken by any program with respect to the size of the input. Time Complexity specifies how the program would behave as the order of size of input is increased. So, Time Complexity is just a function of size of its input. We will go through some of basic and most common time complexities such as:

- *Constant Time Complexity O(1) : constant running time*
- *Linear Time Complexity O(n) : linear running time*
- *Logarithmic Time Complexity O(log n) : logarithmic running time*
- *Log-Linear Time Complexity O(n log n) : log-­linear running time*
- *Polynomial Time Complexity O(n^c) : polynomial running time (c is a constant)*
- *Exponential Time Complexity O(c^n) : exponential running time (c is a constant being raised to a power based on size of input)*

## Constant time Complexity:O(1)
An algorithm is said to have constant time with order O (1) when it is not dependent on the input size n. Irrespective of the input size n, the runtime will always be the same. See example below

```python
# Load the Python libraries
import matplotlib.pyplot as plt
```

```python
# Prob_1

def add_two_numbers(a, b):
    return a + b

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    add_two_numbers(n, n + 1)
    time_values.append(1)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(1) Constant Time Complexity for Addition")
plt.show()
```

## Problem 2: Constant Time Complexity O(1) - Even/Odd Check

```python
import matplotlib.pyplot as plt

def check_even_odd(number):
    return "Even" if number % 2 == 0 else "Odd"

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    # The actual 'input size' for check_even_odd is the magnitude of 'number',
    # but for demonstrating O(1), we show that the *operation count* is constant.
    # We'll use a placeholder constant time value for plotting.
    check_even_odd(n)
    time_values.append(1)  # Representing constant time operation

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input Value (n)") # Clarifying that 'n' here is the input value, not size
plt.ylabel("Time (Constant Operations)")
plt.title("O(1) Constant Time Complexity for Even/Odd Check")
plt.yticks([1], ['1 unit'])
plt.grid(True)
plt.show()
```

#Linear Time Complexity O(n) : linear running time
- *An algorithm is said to have a linear time complexity when the running time increases linearly with the length of the input. When the function involves checking all the values in input data, with this order O(n).*

```python
def linear_time_sum(numbers):
    total = 0
    for value in numbers:
        total += value
    return total

n_values = list(range(1, 101))
time_values = []

for n in n_values:
    data = list(range(n))
    linear_time_sum(data)
    time_values.append(n)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n")
plt.ylabel("Time")
plt.title("O(n) Linear Time Complexity")
plt.show()
```

## Problem 3: Linear Time Complexity O(n) - Count Elements Greater Than Zero

```python
import matplotlib.pyplot as plt

def count_greater_than_zero(numbers):
    count = 0
    for num in numbers:
        if num > 0:
            count += 1
    return count

n_values = list(range(1, 101)) # Input sizes from 1 to 100
time_values = []

for n in n_values:
    # Create a list of 'n' elements for testing.
    # The actual content doesn't matter much for complexity, just the length.
    data = list(range(n)) # Example: [0, 1, ..., n-1]
    count_greater_than_zero(data)
    # The time taken is proportional to 'n', the number of elements in 'data'
    time_values.append(n)

plt.figure()
plt.plot(n_values, time_values)
plt.xlabel("Input size n (Number of elements)")
plt.ylabel("Time (Operations proportional to n)")
plt.title("O(n) Linear Time Complexity for Counting Greater Than Zero")
plt.grid(True)
plt.show()
```

#Logarithmic Time Complexity O(log n) : logarithmic running time
- *An algorithm is said to have a logarithmic time complexity when it reduces the size of the input data in each step. This indicates that the number of operations is not the same as the input size. The number of operations gets reduced as the input size increases. Algorithms are found in binary tree or binary search functions. This involves the search of a given value in an array by splitting the array into two and starting searching in one split. This ensures the operation is not done on every element of the data.*

## Problem 4: Logarithmic Time Complexity O(log n) - Binary Search

```python
import matplotlib.pyplot as plt
import math
import time

def binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    steps = 0 # To count operations, which is proportional to log n

    while low <= high:
        steps += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return steps
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return steps # Target not found, return total steps taken

n_values = list(range(10, 1000, 10)) # Input sizes from 10 to 990
time_values = []

for n in n_values:
    # Create a sorted list for binary search
    data = list(range(n))
    # We'll search for the last element, which often takes the most steps
    target = n - 1

    steps_taken = binary_search(data, target)
    time_values.append(steps_taken)

plt.figure()
plt.plot(n_values, time_values, label="Binary Search (Actual Steps)")
# Plot a theoretical log(n) curve for comparison
plt.plot(n_values, [math.log2(n) for n in n_values], linestyle='--', color='red', label="Theoretical O(log n)")

plt.xlabel("Input size n")
plt.ylabel("Time (Number of steps / log n)")
plt.title("O(log n) Logarithmic Time Complexity for Binary Search")
plt.legend()
plt.grid(True)
plt.show()
```

## Problem 5: Log-Linear Time Complexity O(n log n) - Quicksort

```python
import matplotlib.pyplot as plt
import math
import random

def quicksort(arr):
    steps = [0] # Use a list to pass by reference and accumulate steps
    _quicksort_recursive(arr, 0, len(arr) - 1, steps)
    return steps[0]

def _quicksort_recursive(arr, low, high, steps):
    if low < high:
        pi = _partition(arr, low, high, steps)
        _quicksort_recursive(arr, low, pi - 1, steps)
        _quicksort_recursive(arr, pi + 1, high, steps)

def _partition(arr, low, high, steps):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        steps[0] += 1 # Increment step count for each comparison
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    steps[0] += 1 # Increment step count for final pivot placement
    return i + 1

n_values = list(range(10, 500, 10)) # Input sizes from 10 to 490
time_values = []

for n in n_values:
    # Create a list of 'n' random elements for sorting
    data = [random.randint(0, 1000) for _ in range(n)]

    steps_taken = quicksort(data) # Sort a copy to keep original for next iteration if needed
    time_values.append(steps_taken)

plt.figure()
plt.plot(n_values, time_values, label="Quicksort (Actual Steps)")
# Plot a theoretical n*log(n) curve for comparison
# A scaling factor might be needed to make the theoretical curve align better
scaling_factor = max(time_values) / (max(n_values) * math.log2(max(n_values)))
plt.plot(n_values, [n * math.log2(n) * scaling_factor for n in n_values], linestyle='--', color='red', label="Theoretical O(n log n)")

plt.xlabel("Input size n")
plt.ylabel("Time (Number of steps / n log n)")
plt.title("O(n log n) Log-Linear Time Complexity for Quicksort")
plt.legend()
plt.grid(True)
plt.show()
```

## Problem 6: Polynomial Time Complexity O(n^c) - Nested Loops (O(n^2))

```python
import matplotlib.pyplot as plt
import time

def polynomial_time_example(n):
    count = 0
    for i in range(n):
        for j in range(n):
            count += 1
    return count

n_values = list(range(1, 101, 5)) # Input sizes for better visualization of quadratic growth
time_values = []

for n in n_values:
    steps_taken = polynomial_time_example(n)
    time_values.append(steps_taken)

plt.figure()
plt.plot(n_values, time_values, label="O(n^2) Operations")
# Plot a theoretical n^2 curve for comparison
plt.plot(n_values, [n**2 for n in n_values], linestyle='--', color='red', label="Theoretical O(n^2)")

plt.xlabel("Input size n")
plt.ylabel("Time (Number of Operations)")
plt.title("O(n^2) Polynomial Time Complexity")
plt.legend()
plt.grid(True)
plt.show()
```

## Problem 7: Exponential Time Complexity O(c^n) - Recursive Fibonacci

```python
import matplotlib.pyplot as plt
import time

def fibonacci_recursive(n):
    if n <= 1:
        return n
    else:
        return fibonacci_recursive(n-1) + fibonacci_recursive(n-2)

# We cannot use large n_values for exponential complexity as it would take too long.
# Limiting to a small range to make it runnable.
n_values = list(range(1, 16)) # Input sizes from 1 to 15
time_values = []

for n in n_values:
    start_time = time.perf_counter()
    fibonacci_recursive(n)
    end_time = time.perf_counter()
    time_values.append((end_time - start_time) * 1e6) # Convert to microseconds for better scale

plt.figure()
plt.plot(n_values, time_values, label="Recursive Fibonacci (Actual Time)", marker='o')

# Plot a theoretical 2^n curve for comparison (approximation of O(1.618^n))
# Using a base of ~1.618 (golden ratio) for Fibonacci's exponential growth
# A scaling factor is often needed to align the theoretical curve visually
if n_values and time_values:
    max_n = max(n_values)
    max_time = max(time_values)
    # Avoid division by zero for small n
    if max_n > 0 and (1.618**max_n) > 0:
        scaling_factor = max_time / (1.618**max_n)
    else:
        scaling_factor = 1 # Default if calculation fails
    plt.plot(n_values, [(1.618**n) * scaling_factor for n in n_values], linestyle='--', color='red', label="Theoretical O(1.618^n)")

plt.xlabel("Input size n (Fibonacci Number Index)")
plt.ylabel("Time (microseconds)")
plt.title("O(c^n) Exponential Time Complexity for Recursive Fibonacci")
plt.legend()
plt.grid(True)
plt.show()
```

## Problem 8: Plotting All Common Time Complexities for Comparison

```python
import matplotlib.pyplot as plt
import math

# Define functions for each complexity type
def O_1(n): # Constant
    return [1 for _ in n]

def O_log_n(n): # Logarithmic
    # Use log2 for better representation if n starts from 1, avoid log(0)
    return [math.log2(x) if x > 0 else 0 for x in n]

def O_n(n): # Linear
    return n

def O_n_log_n(n): # Log-Linear
    return [x * (math.log2(x) if x > 0 else 0) for x in n]

def O_n_squared(n): # Polynomial (n^2)
    return [x**2 for x in n]

def O_2_pow_n(n): # Exponential (2^n)
    return [2**x for x in n]

# Generate input sizes
n_values = list(range(1, 20)) # Small range for exponential to be visible

# Calculate time values for each complexity
time_O1 = O_1(n_values)
time_Ologn = O_log_n(n_values)
time_On = O_n(n_values)
time_Onlogn = O_n_log_n(n_values)
time_On2 = O_n_squared(n_values)
time_O2n = O_2_pow_n(n_values)

# Plotting
plt.figure(figsize=(12, 8))

plt.plot(n_values, time_O1, label="O(1) - Constant", marker='o', markersize=4)
plt.plot(n_values, time_Ologn, label="O(log n) - Logarithmic", marker='o', markersize=4)
plt.plot(n_values, time_On, label="O(n) - Linear", marker='o', markersize=4)
plt.plot(n_values, time_Onlogn, label="O(n log n) - Log-Linear", marker='o', markersize=4)
plt.plot(n_values, time_On2, label="O(n^2) - Polynomial", marker='o', markersize=4)

# For exponential, we might need to adjust the y-axis limit or plot a smaller range
# to keep other complexities visible, or show it separately.
# For this combined plot, we'll limit the y-axis to see the initial growth.
plt.plot(n_values, time_O2n, label="O(2^n) - Exponential", marker='o', markersize=4)

plt.xlabel("Input Size (n)")
plt.ylabel("Operations / Time Complexity")
plt.title("Comparison of Time Complexities")
plt.legend()
plt.grid(True)
plt.ylim(0, 100) # Limit y-axis to better visualize the lower complexities
plt.annotate('Note: Exponential O(2^n) grows extremely fast and is truncated here for visibility.',
             xy=(0.05, 0.95), xycoords='axes fraction', fontsize=10, color='red')
plt.show()
```

```python
# Write a function to perform quick sort and show that it has Logarithmic Time Complexity O(log n)
```

```python
import random

def quicksort_function(arr):
    # This is the Quicksort function.
    # For a detailed explanation and plot of its O(n log n) complexity,
    # please refer to the solution provided in Problem 5.
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr) // 2]
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
        return quicksort_function(left) + middle + quicksort_function(right)

# Example usage:
unsorted_list = [random.randint(1, 100) for _ in range(15)]
print(f"Unsorted list: {unsorted_list}")
sorted_list = quicksort_function(unsorted_list)
print(f"Sorted list: {sorted_list}")
```
