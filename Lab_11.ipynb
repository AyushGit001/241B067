# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp(graph):
    n = len(graph)
    dp_cost = [[-1]*(1<<n) for _ in range(n)]
    dp_path = [[-1]*(1<<n) for _ in range(n)]

    def visit(city, mask):
        if mask == (1<<n) - 1:
            return graph[city][0]

        if dp_cost[city][mask] != -1:
            return dp_cost[city][mask]

        min_cost = float('inf')
        best_next_city = -1

        for next_city_candidate in range(n):
            if not (mask & (1<<next_city_candidate)):
                current_cost = graph[city][next_city_candidate] + visit(next_city_candidate, mask | (1<<next_city_candidate))
                if current_cost < min_cost:
                    min_cost = current_cost
                    best_next_city = next_city_candidate
        dp_cost[city][mask] = min_cost
        dp_path[city][mask] = best_next_city
        return min_cost
    min_total_cost = visit(0, 1)
    path = [0]
    current_city = 0
    current_mask = 1
    while current_mask != (1<<n) - 1:
        next_city = dp_path[current_city][current_mask]
        if next_city == -1:
            break
        path.append(next_city)
        current_mask |= (1<<next_city)
        current_city = next_city
    path.append(0)
    return min_total_cost, path
```

```python
graph = [
[0, 10, 15, 20],
[10, 0, 35, 25],
[15, 35, 0, 30],
[20, 25, 30, 0]
]

min_cost, path = tsp(graph)
print("Minimum travelling cost:", min_cost)
print("Optimal path:", path)
```

# Exercise
1. Print path of cities visited
2. Take graph input from user
3. Test with different number of cities

```python
def get_user_graph():
    while True:
        try:
            n = int(input("Enter the number of cities: "))
            if n <= 1:
                print("Number of cities must be at least 2.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter an integer.")

    print(f"Enter the {n}x{n} adjacency matrix (distance between cities). Use 0 for distance from a city to itself, and 'inf' or a very large number for no direct path.\nExample for 2 cities: 0 10\n                       10 0")
    graph = []
    for i in range(n):
        while True:
            try:
                row_str = input(f"Enter row {i+1} (space-separated distances): ")
                row = [float(x) if x.lower() != 'inf' else float('inf') for x in row_str.split()]
                if len(row) != n:
                    print(f"Expected {n} distances, but got {len(row)}. Please re-enter.")
                else:
                    graph.append(row)
                    break
            except ValueError:
                print("Invalid input. Please enter numbers or 'inf'.")
    return graph
user_graph = get_user_graph()
min_cost_user, path_user = tsp(user_graph)

print("\n--- User Input Graph Results ---")
print("Minimum travelling cost:", min_cost_user)
print("Optimal path:", path_user)
```
